# Gini Coefficient

Now, let's dive deeper into the **Gini coefficient**. The Gini coefficient is a number between **0 and 1**, where **0** signals perfect equality between groups and **1** signals perfect inequality. The gini coefficient is used to measure the gender inequality between the gender groups, in this case men and women. If we assume the population is 50% male and 50% female, then gender inequality is when each gender deviates from their 0.5 share. Therefore, gender inequality = $|s_F - s_M|$, where $s_F$ (female share) is the number of women with characteristic $Y$ divided by the total population with $Y$, and $s_M$ (male share) is the number of men with $Y$ divided by the total population with $Y$ — assuming the population of genders was equal.

However, since this is typically not the case (there are not equal numbers of men and women), to find the Gini coefficient for characteristic $Y$, we use the formula:

$$
\sum_g |s_g - p_g|
$$

This equation sums over all genders, which in this case are males and females. It takes $s_g$ (gender $g$'s share), where $s_g$ is the number of people in gender $g$ who have characteristic $Y$ divided by the total number of people with $Y$, and subtracts $p_g$ (gender $g$'s population share), where $p_g$ is the number of people in gender $g$ divided by the total population across all genders.

Since shares and populations each make up a whole, they must sum to 1. Therefore, $s_M$ and $s_F$ are complements of each other (i.e. $s_F = 1 - s_M$), and the same applies to $p_F$ and $p_M$.
