### Dictionary Quick Guide:

We first define the following dictionary:

```python
shopping_list = {'apples': 4, 'bananas': 6, 'bread': 1}
```

Then we can use the following dictionary methods:

| Method                  | Description                                      | Example Output                                                |
|-------------------------|--------------------------------------------------|---------------------------------------------------------------|
| `shopping_list.keys()`  | Returns all **keys** in the shopping list.          | `dict_keys(['apples', 'bananas', 'bread'])`                |
| `shopping_list.values()`| Returns all **values** in the shopping list.     | `dict_values([4, 6, 1])`                                      |
| `shopping_list.items()` | Returns **key-value** pairs as tuples.           | `dict_items([('apples', 4), ('bananas', 6), ('bread', 1)])`   |

**Note:** You can iterate through all three different outputs above.
