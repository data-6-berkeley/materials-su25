# Part 1: Role Conflicts

Recall that roles have an associated authority level. The way the model decides which instructions to follow depends on the **chain of command**. It will ignore or override instructions from lower-authority roles if they conflict with higher ones.

Below is the authority ranking from highest to lowest, alongside a table of the roles you can include in your message list and their purposes:

1. Platform  
2. Developer  
3. User  
4. Guideline  
5. No Authority

| **Role**    | **Authority Level** | **Purpose**                                                                        |
| ----------- | ------------------- | ---------------------------------------------------------------------------------- |
| `system`    | **Platform**        | Sets the assistant’s core behavior, rules, and safety boundaries.                  |
| `developer` | **Developer**       | Adds tone, formatting, and style instructions.                                     |
| `user`      | **User**            | Represents the person prompting the model with a question or command.              |
| `assistant` | **No Authority**    | Represents previous replies from the model, used to maintain conversation history. |
| `tool`      | **No Authority**    | Used when the model calls an external tool (e.g., function calling).               |

In this part of the notebook, you will explore what happens when multiple roles give conflicting instructions.

```python
# Adding our OpenAI API key
from openai import OpenAI
API_KEY = "sk-XXX"  # Replace with your actual OpenAI API key
client = OpenAI(api_key = API_KEY)
```

```python
# Function to prompt the model with a message list from lab 9
def prompt(model, message):
    completion = client.chat.completions.create(model=model, messages=message)
    response_json = completion.model_dump_json()
    response_dict = json.loads(response_json)
    return response_dict["choices"][0]["message"]["content"]
```
